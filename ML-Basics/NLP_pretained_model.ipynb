```python
import spacy
en = spacy.blank("en")
```

```python
text = ("Donald trump (born on june 14,1946) is the 45th and curret president of"
        "the United states.Before entering the politics, he was a businessman and the television personality")
print(text)
```

```python
doc_en = en(text)
```

```python
from IPython.display import HTML, display
import tabulate 

tokens = [[token] for token in doc_en]
display(HTML(tabulate.tabulate(tokens, tablefmt='html')))
```
